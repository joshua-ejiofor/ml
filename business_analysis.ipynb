```python
import pandas as pd

# Load the dataset
df = pd.read_csv('https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv')

```

### What are the info about the dataset? 

```python
# General info: data types, non-null counts
print(df.info())

# Statistical summary
print(df.describe())

# First few rows
print(df.head())
```

### What is the average tip per day?

```python
average_tip_per_day = df.groupby('day')['tip'].mean()
print(average_tip_per_day)
```

### Which day had the highest total bill?

```python
total_bill_per_day = df.groupby('day')['total_bill'].sum()
day_with_max_total_bill = total_bill_per_day.idxmax()
print(f"Day with highest total bill: {day_with_max_total_bill}")
print(f"Total bill: {total_bill_per_day[day_with_max_total_bill]}")
```

###  Add a column for tip percentage
df['tip_percent']

```python
df['tip_pct'] = (df['tip'] / df['total_bill']) * 100
print(df[['total_bill', 'tip', 'tip_pct']].head())
```

### Use ChatGPT to add more rows and columns and save as csv 

```python

# Expanded data with more rows and new columns
data = [
    {"Date": "2023-01-01", "Product": "Widget A", "Region": "North", "Sales": 1000, "Discount": 50, "Cost": 700},
    {"Date": "2023-01-01", "Product": "Widget B", "Region": "South", "Sales": 1500, "Discount": 100, "Cost": 1100},
    {"Date": "2023-01-02", "Product": "Widget A", "Region": "South", "Sales": 2000, "Discount": 150, "Cost": 1300},
    {"Date": "2023-01-02", "Product": "Widget C", "Region": "East",  "Sales": 1800, "Discount": 120, "Cost": 1400},
    {"Date": "2023-01-03", "Product": "Widget B", "Region": "North", "Sales": 2200, "Discount": 200, "Cost": 1600},
    {"Date": "2023-01-03", "Product": "Widget A", "Region": "West",  "Sales": 1300, "Discount": 70,  "Cost": 900},
    {"Date": "2023-01-04", "Product": "Widget C", "Region": "South", "Sales": 2500, "Discount": 180, "Cost": 1900},
]

# Create DataFrame
df = pd.DataFrame(data)

# Add a calculated 'Profit' column
df['Profit'] = df['Sales'] - df['Discount'] - df['Cost']

# Save to CSV
df.to_csv('expanded_sales.csv', index=False)

# Print preview
print(df)
```

### Aggregate Sales by Région

```python

# Load the dataset
df = pd.read_csv('expanded_sales.csv', parse_dates=['Date'])

# Group by 'Region' and sum the 'Sales'
sales_by_region = df.groupby('Region')['Sales'].sum()

# Display the result
print(sales_by_region)
```

###  Aggregate sales by product 

```python



# Group by 'Product' and sum the 'Sales'
sales_by_product = df.groupby('Product')['Sales'].sum()

# Display the result
print(sales_by_product)
```

### Create Pivot Table (Region vs. Product)

```python

# Create pivot table: rows = Region, columns = Product, values = Sales
pivot_table = pd.pivot_table(
    df,
    values='Sales',
    index='Region',
    columns='Product',
    aggfunc='sum',
    fill_value=0  # Replace NaN with 0
)

# Display the pivot table
print(pivot_table)
```

###  Add a Profit Column (20% Margin)

```python

# Add Profit column (20% of Sales)
df['Profit'] = df['Sales'] * 0.20

# Round to 2 decimal places (optional)
df['Profit'] = df['Profit'].round(2)

# Preview the updated DataFrame
print(df.head())

# Optionally save the updated CSV
df.to_csv('expanded_sales_with_profit.csv', index=False)
```

```python

# Original sample data
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Department': ['HR', 'IT', 'IT', 'Marketing'],
    'Salary': [60000, 75000, 72000, 68000]
}

df = pd.DataFrame(data)

# Add new rows
new_rows = pd.DataFrame({
    'Name': ['Ethan', 'Fiona', 'George', 'Hannah'],
    'Department': ['Finance', 'HR', 'IT', 'Marketing'],
    'Salary': [70000, 62000, 78000, 69000]
})

df = pd.concat([df, new_rows], ignore_index=True)

# Add new columns with default or calculated values
df['Bonus'] = df['Salary'] * 0.10  # 10% bonus
df['YearsAtCompany'] = [3, 5, 2, 4, 6, 1, 3, 2]

# Save to CSV
df.to_csv('expanded_employees.csv', index=False)

print("Expanded employees.csv created with added rows and columns.")
print(df)
```

### Find the highest-paid employee in each department.

```python

# Read the CSV file
df = pd.read_csv('expanded_employees.csv')

# Find the index of the highest salary in each department
idx = df.groupby('Department')['Salary'].idxmax()

# Use the indices to get the rows of highest-paid employees
highest_paid = df.loc[idx]

print("Highest-paid employee in each department:")
print(highest_paid)
```

### Find the highest-paid employee in each department.

```python

# Read the CSV file
df = pd.read_csv('expanded_employees.csv')

# Group by Department and calculate mean salary
avg_salary = df.groupby('Department')['Salary'].mean().reset_index()

# Rename columns for clarity
avg_salary.columns = ['Department', 'Average Salary']

print("Average salary per department:")
print(avg_salary)
```
